```python
# Overall P(correct) across both left-correct and right-correct trials

profile_labeled = state_choice_profiles.merge(
    state_mapping[
        ["fold", "raw_state", "provisional_label"]
    ],
    on=["fold", "raw_state"],
    how="left"
)

# First average left/right correct-side conditions within each fold
fold_correct = (
    profile_labeled
    .groupby(
        ["fold", "provisional_label", "delay_s"],
        as_index=False
    )
    ["p_choose_correct_direct"]
    .mean()
    .rename(
        columns={
            "p_choose_correct_direct":
            "predicted_p_correct"
        }
    )
)

# Then summarize across five folds
correct_summary = (
    fold_correct
    .groupby(
        ["provisional_label", "delay_s"],
        as_index=False
    )
    ["predicted_p_correct"]
    .agg(["mean", "std"])
    .reset_index()
)

display(correct_summary)

correct_summary.to_csv(
    POST_DIR / "K4_predicted_accuracy_by_state_and_delay.csv",
    index=False
)

fig, ax = plt.subplots(figsize=(8, 5))

state_order = [
    "Higher-rule-L",
    "Higher-rule-R",
    "Lower-rule-L",
    "Lower-rule-R"
]

for state in state_order:
    sub = (
        correct_summary[
            correct_summary["provisional_label"] == state
        ]
        .sort_values("delay_s")
    )

    ax.errorbar(
        sub["delay_s"],
        sub["mean"],
        yerr=sub["std"],
        marker="o",
        capsize=4,
        label=state
    )

ax.axhline(
    0.5,
    linestyle="--",
    linewidth=1
)

ax.set_xlabel("Delay (s)")
ax.set_ylabel("Predicted P(correct)")
ax.set_ylim(0, 1)
ax.set_title(
    "K=4 state-specific predicted DNMS accuracy"
)
ax.legend()

fig.tight_layout()

fig.savefig(
    POST_DIR / "K4_predicted_accuracy_by_state_and_delay.png",
    dpi=300
)

plt.show()
```
